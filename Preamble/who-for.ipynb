# Who are these notes for?

These notes are written primarily for undergraduate students beginning an upper-level course in modern physics. They connect introductory physics to the study of relativity, quantum mechanics, and their applications. Independent learners with the relevant background are also welcome.

## Preparation and learning goals

You should be familiar with introductory mechanics, electricity and magnetism, and waves, and be prepared to use algebra, derivatives, and integrals. Revisit earlier ideas as needed; recognizing when a familiar model applies is part of learning modern physics.

Prior Python experience is helpful, but you will develop your programming skills alongside the coursework. The goal is to connect physical reasoning, mathematical analysis, and computation: explain your approach, evaluate a model's predictions, and decide whether the results make sense. Write explanations that another student—including **future-you**—can follow.

The [Introduction to Python](https://saturnaxis.github.io/ModernPhysics/Preamble/Python-basics.html) and [Introduction to Markdown](https://saturnaxis.github.io/ModernPhysics/Preamble/Markdown-basics.html) provide practical starting points. Return to them as assignments introduce new calculations or formatting needs.

## Working with the notebooks

Read the rendered chapter on the course website, then use the **rocket ship launch menu** to open its notebook in [Google Colab](https://colab.research.google.com/) where available. Save your own copy to Google Drive before editing. Use **Ctrl+S frequently** (Command+S on a Mac) and confirm that saving has finished.

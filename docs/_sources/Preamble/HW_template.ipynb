# Homework Assignments

The homework develops your ability to construct a scientific argument: identify a physical question, explain an approach, present and interpret the results, and draw a supported conclusion. Write so that another student can follow your reasoning and evaluate your answer. That student may be **future-you**, returning to this work weeks or months later to study for an exam or build on an earlier idea. Explain enough that you can reconstruct your reasoning without relying on memory.

## Submission requirements

A complete homework submission consists of **two files**: a Jupyter notebook (`.ipynb`) and a PDF (`.pdf`) of the same completed work. The notebook contains your editable text and executable code; the PDF provides a readable record of your solution, including equations, figures, and relevant code output.

Use the same basename for both files, following `PHYS321_Chpt##_LastName`, where `##` is the **two-digit chapter number** (`01`, `02`, ..., `12`). For example:

- `PHYS321_Chpt02_Quarles.ipynb`
- `PHYS321_Chpt02_Quarles.pdf`

Follow the individual assignment instructions for the assigned problems and deadline.

## Working in Google Colab

You are encouraged to use [Google Colab](https://colab.research.google.com/) to prepare your homework. Start by saving your own copy of the notebook to Google Drive. Colab periodically saves changes to a Drive-backed notebook, but **use Ctrl+S religiously** (Command+S on a Mac), especially after completing a substantial step and before leaving the page. Check that saving has finished; do not assume that an open notebook or a recent edit has already been saved.

### Group work

Do not rely on Google Docs-style simultaneous editing to keep everyone's changes synchronized in Colab. For group assignments, **designate one person to maintain the master notebook**. Each group member should work in their own copy and send that copy to the person maintaining the master version, clearly identifying the problems or sections they contributed. That person combines the contributions, checks the writing and notation for consistency, and runs the assembled notebook from the beginning. Everyone should review the completed master version before submission.

### Preparing the PDF

Before printing or exporting to PDF, run any code cells from the beginning and save the completed notebook. **Close all cell editors and leave text cells in their rendered view.** Here, “close” means exit editing mode; do not collapse or hide the solution, code, or output needed to understand your work.

A common mistake is to print while a Markdown cell is still open for editing, producing two columns with the source on one side and a partial rendered preview on the other. **Do not submit this split editing view.** The PDF should show the formatted explanation and equations.

Open the saved PDF and check that it matches the notebook: all required text, equations, figures, and relevant code output should be present and readable, with no editing panes, clipped content, or missing sections. Submit both the final `.ipynb` and its matching `.pdf`.

## Writing and formatting expectations

Organize each solution around a clear line of reasoning. The following pattern is useful, but ordinary homework problems do not require four labeled sections.

| Part of the argument | Minimum length | Purpose |
| --- | --- | --- |
| Introduction | At least 2 complete sentences | Motivate the analysis: explain the physical issue and why a particular principle, comparison, or calculation is needed to resolve it. Do more than restate the problem. |
| Method | At least 2 complete sentences, with equations or code as needed | Explain and justify the approach, define the relevant quantities, and state the assumptions. Describe how the analysis will answer the physical question. |
| Results and Discussion | At least 3 complete sentences, with equations, figures, or code as needed | Present the important steps and findings, explain their physical meaning, and discuss relevant comparisons or limitations. Connect the results to the question being investigated. |
| Conclusion | At least 2 complete sentences | Summarize the process and its main result, then provide a specific reasonableness check. Explain why the result makes physical sense. |

These are minimum lengths for the written explanation, not targets to meet with filler. Equations, code, figure labels, and the copied problem statement do not replace explanatory sentences. Use additional sentences whenever needed to make the argument complete. The same expectations apply when the parts are written as connected paragraphs without section headings.

The Introduction should explain why the proposed analysis is useful. For example, identifying acceleration as the quantity that enters Newton's second law motivates examining how it transforms between frames. Simply writing “We need to show that Newton's second law is invariant” does not provide that motivation.

The Conclusion should briefly connect what you did to what you found and explain a concrete check on the result. Depending on the problem, that check might involve units, scale, sign, a limiting case, or agreement with an independent estimate. Saying only “the answer is reasonable” is insufficient. In a research paper, a conclusion might also discuss future directions or connections to the literature; neither is required for these homework solutions.

Use connected paragraphs and equations for short analytical problems. Longer investigations may benefit from descriptive headings. Related parts of a problem may share an introduction and method, provided that the answer to each part is easy to identify.

Use these formatting conventions consistently:

- Give the assignment a top-level heading (`#`) identifying the course and chapter, and each problem a second-level heading (`##`) identifying its number.
- Place the **problem statement in bold inside a blockquote**. For a multipart problem, preserve its part labels.
- Explain your reasoning in complete sentences, incorporating equations where they support the argument. Define symbols, justify consequential steps, and develop a symbolic result before substituting numerical values when appropriate.
- Present numerical answers with units and appropriate precision. Label figure axes and explain what each figure demonstrates.


The example is analytical, so it does not require a Python calculation.

# PHYS 321 — Chapter 02

## Problem 1

> **Show that the form of Newton's second law is invariant under a Galilean transformation.**

<span style="color: #1756a9;"><strong>Introduction</strong></span>

Observers in different inertial frames can assign different positions and velocities to the same particle, but Newton's second law relates force to acceleration. We therefore need to examine how acceleration transforms: differentiating the Galilean position transformation will reveal whether a change of frame alters the equation of motion.

<span style="color: #773899;"><strong>Method</strong></span>

Consider an inertial frame $K$ and a second frame $K'$ moving with constant velocity $u$ along their common $x$-axis. Their axes remain parallel, and their origins coincide at $t=0$. We work in one dimension and use the Galilean transformation

$$
x' = x - ut, \qquad t' = t.
$$

Here, $u$ is the relative velocity of the frames, while $x$ and $x'$ describe the position of the particle. Within the Newtonian description used here, its mass $m$ is constant and the force describing the same interaction is unchanged between these frames, so $F'=F$. We will transform the acceleration and express the equation of motion in the primed coordinates.

<span style="color: #995000;"><strong>Results and Discussion</strong></span>

Because the two frames use the same time coordinate, differentiating the position transformation gives

$$
\frac{dx'}{dt'} = \frac{dx}{dt} - u.
$$

The observers therefore measure different particle velocities. Differentiating again gives

$$
\frac{d^2x'}{dt'^2} = \frac{d^2x}{dt^2} - \frac{du}{dt} = \frac{d^2x}{dt^2},
$$

where the last equality follows because $u$ is constant. Thus, both observers measure the same acceleration, $a'=a$.

Newton's second law in frame $K$ is

$$
F = m\frac{d^2x}{dt^2}.
$$

Substituting the transformed acceleration and using $F'=F$ gives

$$
F' = m\frac{d^2x'}{dt'^2}.
$$

This equation has the same form as the original equation of motion. The result also gives a useful consistency check: when $u=0$, the two coordinate descriptions coincide. If the frame velocity were changing, the acceleration transformation would contain the additional term $-dU/dt$, and the primed frame would require an inertial-force term to retain the form of Newton's second law.

<span style="color: #236b3b;"><strong>Conclusion</strong></span>

Differentiating the Galilean position transformation twice showed that the two inertial frames measure the same particle acceleration. With the same mass and Newtonian force, this gives the same form of Newton's second law in both frames. This result is reasonable because a constant offset in velocity does not change its rate of change; in the limiting case $u=0$, the coordinate descriptions also coincide exactly.
